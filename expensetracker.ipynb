```python
# Expense Tracker System

FILE_NAME = "expenses.txt"

def add_expense():
    try:
        category = input("Enter category (Food/Travel/etc): ")
        amount = float(input("Enter amount: "))

        with open(FILE_NAME, "a") as f:
            f.write(f"{category},{amount}\n")

        print("Expense added successfully!")

    except ValueError:
        print("Invalid amount! Please enter a number.")


def view_expenses():
    try:
        with open(FILE_NAME, "r") as f:
            print("\n--- All Expenses ---")
            for line in f:
                category, amount = line.strip().split(",")
                print(f"Category: {category} | Amount: {amount}")

    except FileNotFoundError:
        print("No expenses found! File does not exist.")



def calculate_total():
    total = 0
    try:
        with open(FILE_NAME, "r") as f:
            for line in f:
                _, amount = line.strip().split(",")
                total += float(amount)

        print("Total Expense:", total)

    except FileNotFoundError:
        print("No data to calculate!")


def category_count():
    counts = {}
    try:
        with open(FILE_NAME, "r") as f:
            for line in f:
                category, _ = line.strip().split(",")
                counts[category] = counts.get(category, 0) + 1

        print("\n--- Category Count ---")
        for cat, count in counts.items():
            print(cat, ":", count)

    except FileNotFoundError:
        print("No data found!")



while True:
    print("\n===== Expense Tracker =====")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Total Expense")
    print("4. Category-wise Count")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_expense()
    elif choice == "2":
        view_expenses()
    elif choice == "3":
        calculate_total()
    elif choice == "4":
        category_count()
    elif choice == "5":
        print("Exiting program...")
        break
    else:
        print("Invalid choice! Try again.")
```
